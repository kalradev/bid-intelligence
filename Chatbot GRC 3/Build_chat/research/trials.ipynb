```python
# Install required packages if not already installed
import sys
import subprocess
import importlib

required_packages = {
    'langchain==0.3.26': 'langchain',
    'langchain-community==0.3.26': 'langchain_community',
    'langchain-core>=0.3.75,<0.3.77': 'langchain_core',  # Fixed version for compatibility
    'langchain-openai==0.3.24': 'langchain_openai',
    'langchain-pinecone==0.2.8': 'langchain_pinecone',
    'langchain-text-splitters>=0.3.0': 'langchain_text_splitters',
    'sentence-transformers==4.1.0': 'sentence_transformers',
    'pypdf==5.6.1': 'pypdf',
    'python-dotenv==1.1.0': 'dotenv',
    'pinecone>=5.0.0': 'pinecone',
    'flask==3.1.1': 'flask'
}

for package_name, import_name in required_packages.items():
    try:
        mod = importlib.import_module(import_name)
        # Check version for langchain-core specifically
        if import_name == 'langchain_core':
            version = getattr(mod, '__version__', 'unknown')
            # Check if version is in compatible range (0.3.75 to <0.3.77)
            try:
                from packaging import version as pkg_version
                if not (pkg_version.parse('0.3.75') <= pkg_version.parse(version) < pkg_version.parse('0.3.77')):
                    print(f"⚠️  {package_name} version {version} may be incompatible. Reinstalling...")
                    subprocess.check_call([sys.executable, "-m", "pip", "install", package_name, "--upgrade", "--force-reinstall"], 
                                        stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)
                    print(f"✅ {package_name} reinstalled with compatible version!")
                else:
                    print(f"✅ {package_name} is already installed (version {version}).")
            except:
                print(f"✅ {package_name} is already installed (version {version}).")
        else:
            print(f"✅ {package_name} is already installed.")
    except ImportError:
        print(f"📦 Installing {package_name}...")
        subprocess.check_call([sys.executable, "-m", "pip", "install", package_name], 
                            stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)
        print(f"✅ {package_name} installed successfully!")

print("\n🎉 All packages are ready!")
```

```python
from langchain_community.document_loaders import PyPDFLoader, DirectoryLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
# Extract text from PDF files
def load_pdf_files(data):
    loader = DirectoryLoader(
        data,
        glob="*.pdf",
        loader_cls=PyPDFLoader
    )

    documents = loader.load()
    return documents
```

```python
extracted_data = load_pdf_files("data")
```

```python
from typing import List
from langchain_core.documents import Document

def filter_to_minimal_docs(docs: List[Document]) -> List[Document]:
    """
    Given a list of Document objects, return a new list of Document objects
    containing only 'source' in metadata and the original page_content.
    """
    minimal_docs: List[Document] = []
    for doc in docs:
        src = doc.metadata.get("source")
        minimal_docs.append(
            Document(
                page_content=doc.page_content,
                metadata={"source": src}
            )
        )
    return minimal_docs
```

```python
minimal_docs = filter_to_minimal_docs(extracted_data)
```

```python
minimal_docs
```

```python
# Split the documents into smaller chunks
def text_split(minimal_docs):
    text_splitter = RecursiveCharacterTextSplitter(
        chunk_size=500,
        chunk_overlap=20,
    )
    texts_chunk = text_splitter.split_documents(minimal_docs)
    return texts_chunk
```

```python
texts_chunk = text_split(minimal_docs)
print(f"Number of chunks: {len(texts_chunk)}")
```

```python
from langchain_community.embeddings import HuggingFaceEmbeddings

def download_embeddings():
    """
    Download and return the HuggingFace embeddings model.
    """
    model_name = "sentence-transformers/all-MiniLM-L6-v2"
    embeddings = HuggingFaceEmbeddings(
        model_name=model_name
    )
    return embeddings   

embedding = download_embeddings()
```

```python
from langchain_openai import ChatOpenAI

chatModel = ChatOpenAI(model="gpt-4o")
```

```python
# Import chains with auto-install if needed
import sys
import subprocess

try:
    from langchain.chains import create_retrieval_chain
    from langchain.chains.combine_documents import create_stuff_documents_chain
except ImportError:
    print("📦 Installing langchain...")
    subprocess.check_call([sys.executable, "-m", "pip", "install", "langchain==0.3.26"])
    from langchain.chains import create_retrieval_chain
    from langchain.chains.combine_documents import create_stuff_documents_chain

from langchain_core.prompts import ChatPromptTemplate
print("✅ All imports successful!")
```

```python
system_prompt = (
    "You are a GRC (Governance, Risk, and Compliance) assistant for question-answering tasks. "
    "Use the following pieces of retrieved context from the Sanchlan GRC knowledge base to answer "
    "the question. If you don't know the answer, say that you "
    "don't know. Use three sentences maximum and keep the "
    "answer concise."
    "\n\n"
    "{context}"
)


prompt = ChatPromptTemplate.from_messages(
    [
        ("system", system_prompt),
        ("human", "{input}"),
    ]
)
```
